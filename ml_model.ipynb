```python
data = pd.DataFrame(data)
data #to convert data into tabular form (use pandas dataframe)
```

```python
#select features => input instances that effect the output
X = data[
    ["CGPA","Skills","Projects","Internships","Certifications","DSA_Score"]
]
X
```

```python
#select target => output that we need
Y = data["Placed"]
Y
```

```python
#variables by usind train_test_split
X_train , X_test , Y_train , Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
print("Training students:", len(X_train))
print("Testing students:", len(X_test))
```

```python
#creating the ml model
model = DecisionTreeClassifier(random_state=42)
```

```python
#train the model by .fit
model.fit(X_train, Y_train)
print('\nModel is trained')
```

```python
#test the model by .predict
predictions = model.predict(X_test)

print("\nActual values:",list(Y_test))
print("\nPredicted values:",list(predictions))
```

```python
#testing accuracy of model
accuracy = accuracy_score(Y_test,predictions)
print(accuracy)
```
